### Objetivo

Comparar **SVR con diferentes kernels** (linear, RBF, polynomial) contra una regresion lineal. Se usa un dataset inline con relacion no lineal para demostrar cuando SVR con kernel RBF supera a un modelo lineal.

### Contexto de negocio

**El cliente:** un e-commerce de tecnologia quiere predecir el gasto anual de cada cliente.

**El problema:** la relacion entre fidelidad y gasto no es lineal — los clientes nuevos y los muy fieles gastan mucho, pero los del medio gastan poco. Una regresion lineal no captura esto.

**La pregunta:** SVR con kernel RBF puede capturar esta relacion no lineal?

## 4. Preparacion de datos

**Estandarizacion obligatoria para SVR.** Los kernels SVM son sensibles a la escala de las features.

```python
X = data[[
    "Score_Fidelidad",
    "Ingresos_Mensuales"
]]

y = data["Compra_iPhone"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    stratify=y,
    random_state=42
)

scaler_X = StandardScaler()

X_train_s = scaler_X.fit_transform(X_train)
X_test_s = scaler_X.transform(X_test)

print(f"Train: {X_train_s.shape[0]} | Test: {X_test_s.shape[0]}")
```

- Aplicar estratificado ya que es un ejercicio de clasificación.
- Aplicar normalización ya que disponemos variables de diferentes rangos. 

## 5. Modelo base — SVM lineal

```python
modelo_linear = SVC(
    kernel="linear",
    C=1.0,
    random_state=42
)

modelo_linear.fit(
    X_train_s,
    y_train
)

y_pred_linear = modelo_linear.predict(X_test_s)

print(
    classification_report(
        y_test,
        y_pred_linear,
        target_names=["No compra (0)", "Compra (1)"]
    )
)
```

### SVM — Kernel Polinómico

```python
# Construcción del modelo SVM con kernel polinomial

modelo_poly = SVC(
    kernel="poly",
    degree=2,
    C=1.0,
    gamma="scale",
    coef0=1,
    random_state=42
)

modelo_poly.fit(
    X_train_s,
    y_train
)

y_pred_poly = modelo_poly.predict(X_test_s)

print(
    classification_report(
        y_test,
        y_pred_poly,
        target_names=["No compra (0)", "Compra (1)"]
    )
)
```

#### SVM — Kernel RBF

## Conclusión

El análisis muestra que la relación entre **ingresos, fidelidad y compra de iPhone no se separa completamente mediante una frontera lineal**.

El **SVM lineal alcanza un 93,2% de accuracy**, un resultado sólido, pero el **kernel RBF mejora hasta el 96,6%**, capturando mejor la estructura no lineal de los clientes.

El kernel polinómico, en cambio, obtiene un rendimiento claramente inferior, con un 62,7% de accuracy, por lo que no aporta una ventaja en este caso.


### Insight de negocio

**El perfil de compra no depende únicamente de los ingresos. La combinación de capacidad económica y fidelidad tecnológica permite identificar mejor a los clientes con mayor propensión a comprar un iPhone.**

Por tanto, para este dataset, **RBF es la opción más adecuada**, aunque el modelo debe considerarse ilustrativo porque se trabaja con un dataset sintético y únicamente dos variables predictoras.
