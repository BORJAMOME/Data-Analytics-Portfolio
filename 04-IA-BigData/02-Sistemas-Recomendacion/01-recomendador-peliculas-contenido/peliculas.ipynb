# Sistemas de recomendación basado en contenido
## Recomendador de peliculas

```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
ruta = r"C:\Users\borja\Desktop\Github\Data-Analytics-Portfolio\04-IA-BigData\02-Sistemas-Recomendacion\01-recomendador-peliculas-contenido\data\peliculas_netflix.xlsx"

df = pd.read_excel(ruta)
```

```python
df.head(20)
```

```python
#Sistema de recomendación basado en la descripción. Transformación del texto en numeros con la liberia a aplicar. 

vectorizer = TfidfVectorizer()
matriz_tfid = vectorizer.fit_transform(
    df["descripcion"]
)
```

```python
vectorizer.get_feature_names_out()
```

```python
#Crear matriz de similitud 

matriz_similitud = cosine_similarity(
    matriz_tfid
)
matriz_similitud
```

```python
#Convertir la matriz a un dataframe

df_similitud = pd.DataFrame(
    matriz_similitud,
    index=df["titulo"],
    columns=df["titulo"]
)

df_similitud.round(2)
```

```python
#Construir el recomendador

pelicula_seleccionada = "Up"

similitudes = df_similitud[pelicula_seleccionada]

similitudes
```

```python
recomendaciones = (
    similitudes.sort_values(ascending=False)
)
recomendaciones
```

```python
top_5 = recomendaciones.head(6)
top_5
```
